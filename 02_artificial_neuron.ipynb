## 1. Biological Neuron
* Definition: The fundamental unit of the brain and nervous system. It receives electrical impulses (signals) from other neurons through dendrites, processes them in the cell body (soma), and transmits an output signal through the axon if the threshold is met.
* Use Cases: Studying cognitive neuroscience. Inspiration for brain-inspired computing and SNNs (Spiking Neural Networks).
* Pros:
    * Highly efficient and parallelized processing.
    * Capable of complex learning and adaptation (neuroplasticity).
* Cons:
    * Extremely complex and difficult to model precisely in silicon.
    * Not directly efficient for traditional mathematical optimization tasks.

## 2. Artificial Neuron (Perceptron)
* Definition: A mathematical function conceived as a model of biological neurons. It takes multiple inputs, multiplies them by weights, adds a bias, and passes the result through an activation function to produce an output.
* Use Cases:
    * Building block for Deep Learning architectures. Pattern recognition, classification, and regression tasks.
* Pros:
    * Differentiable, allowing for training via backpropagation.
    * Can model complex, non-linear relationships when combined in layers.
* Cons:
    * Requires large datasets for training.
    * Can be a "black box" regarding interpretability compared to traditional statistics.
### Components of an Artificial Neuron
* Input ($x$): The raw data or signal fed into the neuron.
* Weight ($w$): Determines the strength of the connection between inputs. It dictates how much influence an input has on the output.
* Bias ($b$): An additional parameter that allows the neuron to shift the activation function to the left or right, independent of the input.
* Weighted Sum ($z$): The linear combination of inputs and weights plus bias ($z = \sum(w \cdot x) + b$).
* Activation Function ($f$): A non-linear function (e.g., Sigmoid, ReLU, Tanh) that determines whether and to what extent the neuron should "fire" (pass the signal forward).
*Output ($y$): The final result produced by the neuron ($y = f(z)$).

### Visualization: The Anatomy of a Neuron
The code below visualizes the transformation of an input through an activation function.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Activation Function (e.g., Sigmoid)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 2. Generate data for weighted sum (z)
z = np.linspace(-10, 10, 100)
output = sigmoid(z)

# 3. Visualize the Neuron's Activation
plt.figure(figsize=(8, 4))
plt.plot(z, output, label="Sigmoid Activation", color='blue', lw=2)
plt.axvline(x=0, color='red', linestyle='--', alpha=0.5, label="Threshold (Bias Shift)")
plt.title("Neuron Activation (Sigmoid)")
plt.xlabel("Weighted Sum (z)")
plt.ylabel("Output (y)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```

| Component | Function | Role |
| :--- | :--- | :--- |
| **Input** | Data Source | Provides the feature information. |
| **Weight** | Multiplier | Controls signal importance/sensitivity. |
| **Bias** | Offset | Provides flexibility in learning boundaries. |
| **Sum** | Aggregator | Combines weighted inputs linearly. |
| **Activation** | Transformer | Introduces non-linearity to the model. |
| **Output** | Result | Represents the prediction or firing state. |
