## 1. Perceptron
* Definition: A single-layer binary classifier that computes a weighted sum of inputs and passes it through a step function to make a prediction.
* Use Cases:
    * Binary classification tasks.
    * Solving linear separation problems (e.g., logical AND/OR gates).
* Pros:
    * Simple to implement and computationally lightweight.
    * Foundational stepping stone for modern deep learning.
* Cons:
    * Cannot solve non-linearly separable problems (e.g., the XOR problem).
    * Uses a step activation function which is non-differentiable, preventing standard gradient-based optimization.

## 2. Multi Layer Perceptron (MLP)
* Definition: A class of feedforward artificial neural network consisting of at least three layers of nodes (input, hidden, output) utilizing non-linear activation functions to handle complex mappings.
* Use Cases:
    * Complex pattern recognition and classification.
    * Non-linear regression tasks.
* Pros:
    * Universal Approximation Theorem: Can approximate any continuous function given enough hidden units.
    * Capable of learning complex, non-linear relationships via backpropagation.
* Cons:
    * Prone to overfitting if the model is too complex or data is scarce.
    * Computationally expensive to train with deep architectures.

## 3. Input Layer
* Definition: The initial layer of a neural network that receives raw feature data from the dataset without performing any mathematical transformations.
* Use Cases:
    * Ingesting tabular features, image pixels, or sequence vectors into the network.
* Pros:
    * Directly bridges raw data dimensions with the neural architecture.
    * No trainable parameters, reducing early computational overhead.
* Cons:
    * Cannot learn abstract features or filter noise by itself.

## 4. Hidden Layer
* Definition: Intermediate layers situated between the input and output layers that extract hierarchical features and abstract representations from the data.
* Use Cases:
    * Feature extraction in deep learning, computer vision, and NLP.
* Pros:
    * Enables the network to model highly complex, multi-level data interactions.
    * Automates feature engineering.
* Cons:
    * Deep hidden layers can suffer from vanishing or exploding gradients.
    * Increases model "black-box" opacity.

## 5. Output Layer
* Definition: The final layer of a neural network that produces the network's ultimate prediction, classification probability, or regression value.
* Use Cases:
    * Multi-class classification (via Softmax) or continuous forecasting.
* Pros:
    * Translates internal representations into actionable, task-specific predictions.
    * Flexible design depending on the objective function.
* Cons:
    * Strictly constrained by the task type and chosen loss function.

## 6. Fully Connected Layer (Dense Layer)
* Definition: A layer where each neuron is connected to every neuron in the previous layer, computing linear combinations followed by non-linear activations.
* Use Cases:
    * Final decision-making stages of classification and regression networks.
* Pros:
    * Highly expressive, capturing global interactions across all extracted features.
* Cons:
    * Extremely high parameter count, leading to massive memory usage and vulnerability to overfitting.

## 7. Network Architecture
* Definition: The structural layout of a neural network, encompassing the number of layers, number of neurons per layer, layer types (Dense, Convolutional, etc.), and overall connectivity pattern.
* Use Cases:
    * Designing tailored network topologies for specific data modalities (e.g., CNNs for images, RNNs for text).
* Pros:
    * Determines the model's inductive bias and capacity to solve specific problems.
* Cons:
    * Finding an optimal architecture often requires extensive trial-and-error or automated architecture search.

## 8. Model Parameters
* Definition: The internal configuration variables of a neural network (weights and biases) whose values are optimized iteratively via training algorithms.
* Use Cases:
    * Encoding and storing the learned patterns of a dataset.
* Pros:
    * Allows models to generalize from training instances to unseen test data.
* Cons:
    * Large parameter spaces require significant hardware resources and heavy regularization strategies.

### Visualizing MLP Decision Boundaries
The code below trains a Multi-Layer Perceptron (MLP) on a non-linear dataset (moons) and visualizes how hidden layers allow the network to draw complex decision boundaries.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.neural_network import MLPClassifier
from sklearn.inspection import DecisionBoundaryDisplay

# 1. Generate non-linear synthetic dataset
X, y = make_moons(n_samples=200, noise=0.2, random_state=42)

# 2. Define and train MLP Classifier
clf = MLPClassifier(hidden_layer_sizes=(10, 10), activation='relu', max_iter=2000, random_state=42)
clf.fit(X, y)

# 3. Plotting the Decision Boundary correctly
plt.figure(figsize=(8, 6))
disp = DecisionBoundaryDisplay.from_estimator(
    clf, X, response_method="predict", cmap=plt.cm.coolwarm, alpha=0.8
)

# Plot data points on the same axes
plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.coolwarm, edgecolors='k')
plt.title("MLP (Neural Network) Decision Boundary on Non-Linear Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```
