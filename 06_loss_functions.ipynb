# Regression Losses

## 1. Mean Squared Error (MSE)
* Definition: The average squared difference between estimated values and the actual target.
* Use Cases:
    * Standard regression tasks.
* Pros:
    * Smooth, differentiable everywhere; heavily penalizes large errors.
* Cons:
    * Extremely sensitive to outliers.

### Visualizing Mean Squared Error (MSE)

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
mse = errors ** 2

plt.figure(figsize=(6, 3.5))
plt.plot(errors, mse, color='blue', linewidth=2)
plt.title(r"Mean Squared Error (MSE)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 2. Root Mean Squared Error (RMSE)
* Definition: The square root of the Mean Squared Error.
* Use Cases:
    * Regression where loss must match the target variable's scale.
* Pros:
    * Intuitive because it shares original units.
* Cons:
    * Retains MSE's sensitivity to outliers.

### Visualizing Root Mean Squared Error (RMSE)

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
rmse = np.sqrt(errors ** 2)

plt.figure(figsize=(6, 3.5))
plt.plot(errors, rmse, color='darkblue', linewidth=2)
plt.title("Root Mean Squared Error (RMSE)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 3. Mean Absolute Error (MAE)
* Definition: The average absolute magnitude of errors.
* Use Cases:
    * Regression tasks with noisy data containing outliers.
* Pros:
    * Robust to outliers.
* Cons:
    * Non-differentiable at zero.

### Visualizing Mean Absolute Error (MAE)

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
mae = np.abs(errors)

plt.figure(figsize=(6, 3.5))
plt.plot(errors, mae, color='green', linewidth=2)
plt.title("Mean Absolute Error (MAE)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 4. Huber Loss
* Definition: Combines MSE near zero and MAE for large errors.
* Use Cases:
    * Robust regression requiring outlier resistance.
* Pros:
    * Resistant to outliers while remaining smooth.
* Cons:
    * Requires tuning a threshold parameter ($\delta$).

### Visualizing Huber Loss

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
delta = 1.0
huber = np.where(np.abs(errors) <= delta, 0.5 * errors**2, delta * (np.abs(errors) - 0.5 * delta))

plt.figure(figsize=(6, 3.5))
plt.plot(errors, huber, color='red', linewidth=2)
plt.title(r"Huber Loss ($\delta=1.0$)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

# Classification Losses

## 1. Binary Cross Entropy (BCE)
* Definition: Measures performance for models outputting probabilities between 0 and 1.
* Use Cases:
    * Binary classification tasks.
* Pros:
    * Strongly penalizes confident, incorrect predictions.
* Cons:
    * Prone to numerical instability near 0 or 1.6. 

### Visualizing Binary Cross Entropy (BCE)

```python
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0.01, 0.99, 100)
bce = -np.log(p) # For true label y = 1

plt.figure(figsize=(6, 3.5))
plt.plot(p, bce, color='purple', linewidth=2)
plt.title("Binary Cross Entropy (True Label y=1)", fontsize=11, fontweight='bold')
plt.xlabel(r"Predicted Probability ($p$)")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 2. Categorical Cross Entropy (CCE)
* Definition: Measures divergence between one-hot encoded labels and predicted probabilities.
* Use Cases:
    * Multi-class classification.
* Pros:
    * Pairs naturally with the Softmax activation function.
* Cons:
    * Requires memory-heavy one-hot encoding.

### Visualizing Categorical Cross Entropy (CCE)

```python
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0.01, 0.99, 100)
cce = -np.log(p)

plt.figure(figsize=(6, 3.5))
plt.plot(p, cce, color='magenta', linewidth=2)
plt.title("Categorical Cross Entropy", fontsize=11, fontweight='bold')
plt.xlabel("Probability for True Class")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 3. Sparse Categorical Cross Entropy
* Definition: Cross entropy loss accepting integer-encoded class labels.
* Use Cases:
    * Multi-class classification with large label spaces.
* Pros:
    * Saves memory by avoiding one-hot encoding.
* Cons:
    * Requires proper integer index alignment.

### Visualizing Sparse Categorical Cross Entropy

```python
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0.01, 0.99, 100)
sparse_cce = -np.log(p)

plt.figure(figsize=(6, 3.5))
plt.plot(p, sparse_cce, color='deeppink', linewidth=2, linestyle='--')
plt.title("Sparse Categorical Cross Entropy", fontsize=11, fontweight='bold')
plt.xlabel("Probability for True Class")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 4. Focal Loss
* Definition: Dynamically scaled cross-entropy that down-weights easy-to-classify examples.
* Use Cases:
    * Highly imbalanced classification datasets.
* Pros:
    * Effectively handles severe class imbalance.
* Cons:
    * Introduces extra hyperparameters requiring careful tuning.

### Visualizing Focal Loss

```python
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0.01, 0.99, 100)
gamma = 2.0
focal = -((1 - p) ** gamma) * np.log(p)

plt.figure(figsize=(6, 3.5))
plt.plot(p, focal, color='orange', linewidth=2)
plt.title(r"Focal Loss ($\gamma=2.0$)", fontsize=11, fontweight='bold')
plt.xlabel(r"Predicted Probability ($p$)")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
