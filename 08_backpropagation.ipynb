## 1. Chain Rule
* Definition: A fundamental calculus rule used to compute the derivative of composite functions ($f(g(x))' = f'(g(x)) \cdot g'(x)$).
* Use Cases:
    * Calculating how changes in downstream layers affect upstream parameters across deep architectures.
* Pros:
    * Enables efficient local gradient computation by breaking down complex global dependencies into sequential multiplications.
* Cons:
    * Multiplying many fractional derivatives across deep networks can cause vanishing gradients.

### Visualizing the Chain Rule Concept

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 100)
# Composite function y = (x^2 + 1)^2
y = (x**2 + 1)**2
# Derivative via chain rule: dy/dx = 2*(x^2 + 1) * 2x
dy_dx = 4 * x * (x**2 + 1)

plt.figure(figsize=(6, 3.5))
plt.plot(x, y, label=r"$y = (x^2 + 1)^2$", color='royalblue', linewidth=2)
plt.plot(x, dy_dx, label=r"Derivative $\frac{dy}{dx}$", color='darkorange', linewidth=2, linestyle='--')
plt.title("Chain Rule Function and Derivative", fontsize=11, fontweight='bold')
plt.xlabel("x")
plt.ylabel("Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

## 2. Gradient Calculation
* Definition: The computation of partial derivatives of the loss function with respect to every individual weight and bias in the network.
* Use Cases:
    * Determining the slope and magnitude of error sensitivity for parameter adjustments.
* Pros:
    * Provides precise mathematical insight into how each weight influences the final output error.
* Cons:
    * Computationally expensive and memory-intensive for large neural network structures.

### Visualizing Gradient Flow across Layers

```python
import matplotlib.pyplot as plt

layers = ['Input', 'Hidden 1', 'Hidden 2', 'Output']
gradients = [0.01, 0.05, 0.2, 0.8]  # Simulated gradient magnitudes flowing backward

plt.figure(figsize=(6, 3.5))
plt.bar(layers, gradients, color='purple', alpha=0.7, edgecolor='black')
plt.title("Gradient Magnitude Backward Flow", fontsize=11, fontweight='bold')
plt.xlabel("Network Layers")
plt.ylabel("Average Gradient Magnitude")
plt.grid(True, axis='y', linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 3. Error Propagation
* Definition: The backward pass mechanism where the error calculated at the output layer is distributed and passed back through hidden layers.
* Use Cases:
    * Assigning blame for final errors to individual internal neurons.
* Pros:
    * Automatically routes error signals to every parameter based on its architectural connectivity.
* Cons:
    * Vulnerable to vanishing or exploding gradient anomalies over long propagation chains.

### Visualizing Error Signal Decay (Vanishing Gradient)

```python
import numpy as np
import matplotlib.pyplot as plt

layer_depth = np.arange(1, 11)
# Simulating exponential decay of gradients over depth
vanishing_grad = 0.9 ** layer_depth

plt.figure(figsize=(6, 3.5))
plt.plot(layer_depth, vanishing_grad, marker='o', color='crimson', linewidth=2)
plt.title("Error Signal Decay Across Deep Layers", fontsize=11, fontweight='bold')
plt.xlabel("Layer Depth (Distance from Output)")
plt.ylabel("Gradient Strength")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 4. Weight Update
* Definition: The optimization step where network parameters are modified using the calculated gradients and learning rate ($w = w - \alpha \cdot \frac{\partial L}{\partial w}$).
* Use Cases:
    * Iteratively minimizing the loss function to improve model predictions.
* Pros:
    * Drives the network toward optimal parameter states, enabling learning from data.
* Cons:
    * Suboptimal learning rate selection can cause training divergence or trapping in local minima.

### Visualizing Weight Step Convergence

```python
import numpy as np
import matplotlib.pyplot as plt

iterations = np.arange(1, 15)
# Simulating weight convergence toward optimal value (e.g., 0)
weights = 3.0 * (0.7 ** (iterations - 1))

plt.figure(figsize=(6, 3.5))
plt.plot(iterations, weights, marker='o', color='forestgreen', linewidth=2)
plt.title("Weight Convergence During Training Updates", fontsize=11, fontweight='bold')
plt.xlabel("Training Iteration")
plt.ylabel("Weight Value ($w$)")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 5. Computational Graph
* Definition: A directed graph that models mathematical expressions where nodes represent operations and edges represent data arrays.
* Use Cases:
    * Structuring and organizing complex neural network expressions for modular evaluation.
* Pros:
    * Simplifies automatic differentiation by breaking global equations into manageable local operations.
* Cons:
    * Requires caching forward activation values in memory to compute backward passes.

### Visualizing a Simple Computational Graph Flow

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3))
ax.axis('off')

# Draw nodes representing a simple graph: x -> (*) -> (+) -> Loss
ax.text(0.1, 0.5, r"Input ($x, w$)", fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="lightblue"))
ax.text(0.4, 0.5, r"Multiply" + "\n" + r"($z = w \cdot x$)", fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="lightyellow"))
ax.text(0.7, 0.5, r"Activation" + "\n" + r"($a = f(z)$)", fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="lightgreen"))

ax.annotate('', xy=(0.35, 0.5), xytext=(0.25, 0.5), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate('', xy=(0.65, 0.5), xytext=(0.55, 0.5), arrowprops=dict(arrowstyle="->", lw=1.5))

plt.title("Computational Graph Forward Flow", fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 6. Automatic Differentiation (AutoDiff)
* Definition: A computational technique used by software frameworks to evaluate exact derivatives of functions specified by computer code.
* Use Cases:
    * Powering backend gradient calculations in deep learning libraries like PyTorch and TensorFlow.
* Pros:
    * Eliminates manual calculus derivation errors and automates gradient tracking.
* Cons:
    * Adds memory and framework overhead to track operational histories during execution.

### Visualizing AutoDiff Node Tracking Concept

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3))
ax.axis('off')

ax.text(0.15, 0.6, "Forward Pass:\nRecord Operations", fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="lightcyan"))
ax.text(0.55, 0.6, "Backward Pass:\nApply Chain Rule", fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="peachpuff"))

ax.annotate('', xy=(0.52, 0.6), xytext=(0.42, 0.6), arrowprops=dict(arrowstyle="->", lw=1.5, color='red'))

plt.title("Automatic Differentiation Lifecycle", fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```
