## 1. Dataset
* Definition: A complete collection of data records or observations used to train, validate, and test machine learning models.
* Use Cases:
    * The primary input foundation for any machine learning or deep learning project.
* Pros:
    * Contains the broad statistical distribution and patterns needed for generalized learning.
* Cons:
    * Can suffer from quality issues such as class imbalance, noise, and hidden biases.

### Visualizing a Dataset Structure

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.random.rand(100)
y = np.random.rand(100)

plt.figure(figsize=(6, 3.5))
plt.scatter(x, y, color='royalblue', alpha=0.7, edgecolor='black')
plt.title("Full Dataset Distribution", fontsize=11, fontweight='bold')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 2. Sample (Instance)
* Definition: A single data point or row within a dataset, consisting of input features and an optional target label.
* Use Cases:
    * Individual inference, online learning, or single-instance evaluation.
* Pros:
    * Represents the atomic, real-world unit of observation.
* Cons:
    * A single sample is often noisy and insufficient on its own to train a generalized model.

### Visualizing a Single Sample Highlighted

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.random.rand(50)
y = np.random.rand(50)

plt.figure(figsize=(6, 3.5))
plt.scatter(x, y, color='lightgray', edgecolor='black', label="Dataset Samples")
# Highlight one sample
plt.scatter([x[10]], [y[10]], color='red', s=120, zorder=5, label="Single Sample")
plt.title("A Single Sample in Context", fontsize=11, fontweight='bold')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Batch
* Definition: The entire dataset processed together in a single optimization step (forward and backward pass).
* Use Cases:
    * Small datasets or batch gradient descent optimization where exact gradient calculation is viable.
* Pros:
    * Provides a smooth, noise-free gradient trajectory directly toward the minimum.
* Cons:
    * Computationally impossible and memory-prohibitive for massive modern datasets.

## 4. Mini-Batch
* Definition: A small, randomly selected subset of the dataset processed together during a single training step.
* Use Cases:
    * The industry standard for training deep learning models across all modern frameworks.
* Pros:
    * Balances memory efficiency, computational speed, and stable gradient updates.
* Cons:
    * Introduces minor gradient noise, requiring careful tuning of the batch size.

## 5. Epoch
* Definition: One complete pass of the entire dataset through the training algorithm (all mini-batches processed once).
* Use Cases:
    * Tracking overall model training progress and scheduling learning rate decay across training sessions.
* Pros:
    * Ensures the model sees every data point multiple times to deeply learn underlying patterns.
* Cons:
    * Too few epochs cause underfitting; too many epochs lead to severe overfitting.

### Visualizing Epoch Progress (Loss over Epochs)

```python
import numpy as np
import matplotlib.pyplot as plt

epochs = np.arange(1, 21)
train_loss = np.exp(-epochs / 5.0) + 0.1

plt.figure(figsize=(6, 3.5))
plt.plot(epochs, train_loss, marker='o', color='purple', linewidth=2)
plt.title("Training Loss Across Multiple Epochs", fontsize=11, fontweight='bold')
plt.xlabel("Epochs")
plt.ylabel("Loss Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 6. Iteration
* Definition: A single parameter update step performed using one mini-batch of data.
* Use Cases:
    * Granular measurement of progress within an individual training epoch.
* Pros:
    * Frequent parameter updates allow the model to adapt and learn rapidly.
* Cons:
    * The number of iterations per epoch grows as batch size shrinks, increasing training logging overhead.
