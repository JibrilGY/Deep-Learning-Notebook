# Classification Metrics

## 1. Accuracy
* Definition: The ratio of correct predictions to the total number of predictions made.
* Use Cases:
    * General classification tasks with balanced class distributions.
* Pros:
    * Highly intuitive, easy to compute, and simple to interpret.
* Cons:
    * Highly misleading when applied to severely imbalanced datasets.

### Visualizing Accuracy Metric

```python
import matplotlib.pyplot as plt

categories = ['Correct Predictions', 'Incorrect Predictions']
values = [92, 8]

plt.figure(figsize=(6, 3.5))
plt.bar(categories, values, color=['forestgreen', 'crimson'], alpha=0.8, edgecolor='black')
plt.ylim(0, 100)
plt.title("Accuracy Breakdown (92%)", fontsize=11, fontweight='bold')
plt.ylabel("Percentage (%)")
plt.grid(True, axis='y', linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 2. Precision
* Definition: The ratio of true positive predictions to total predicted positives ($\frac{TP}{TP + FP}$).
* Use Cases:
    * Scenarios where false positives are extremely costly (e.g., spam email classification).
* Pros:
    * Measures the exactness and reliability of positive predictions.
* Cons:
    * Completely ignores false negatives and actual missed positive cases.

## 3. Recall (Sensitivity)
* Definition: The ratio of true positives to total actual positive instances in the dataset ($\frac{TP}{TP + FN}$).
* Use Cases:
    * Critical domains like medical diagnosis where missing a disease is dangerous.
* Pros:
    * Measures completeness and ability to capture all positive cases.
* Cons:
    * Can be easily gamed by predicting everything as positive.

## 4. F1 Score
* Definition: The harmonic mean of precision and recall ($2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$).
* Use Cases:
    * Imbalanced classification tasks requiring a balance between precision and recall.
* Pros:
    * Provides a single balanced metric incorporating both false positives and false negatives.
* Cons:
    * Treats precision and recall with equal importance, which may not suit every domain.

### Visualizing F1 Score Comparison

```python
import matplotlib.pyplot as plt

metrics = ['Precision', 'Recall', 'F1 Score']
scores = [0.88, 0.94, 0.91]

plt.figure(figsize=(6, 3.5))
plt.bar(metrics, scores, color=['royalblue', 'darkorange', 'purple'], alpha=0.8, edgecolor='black')
plt.ylim(0, 1.0)
plt.title("Precision, Recall, and F1 Score", fontsize=11, fontweight='bold')
plt.ylabel("Score")
plt.grid(True, axis='y', linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 5. ROC Curve (Receiver Operating Characteristic)
* Definition: A graphical plot showing the diagnostic ability of a binary classifier across all classification thresholds.
* Use Cases:
    * Evaluating classification threshold trade-offs between True Positive Rate and False Positive Rate.
* Pros:
    * Independent of the classification threshold choice.
* Cons:
    * Can provide an overly optimistic visual performance representation under severe class imbalance.

### Visualizing ROC Curve

```python
import numpy as np
import matplotlib.pyplot as plt

fpr = np.array([0.0, 0.1, 0.2, 0.5, 1.0])
tpr = np.array([0.0, 0.8, 0.9, 0.95, 1.0])

plt.figure(figsize=(6, 3.5))
plt.plot(fpr, tpr, color='purple', linewidth=2, label="ROC Curve")
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label="Random Guess")
plt.title("ROC Curve Analysis", fontsize=11, fontweight='bold')
plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. AUC (Area Under the ROC Curve)
* Definition: The scalar measure of the entire area underneath the ROC curve.
* Use Cases:
    * Model comparison and overall ranking performance assessment.
* Pros:
    * Provides a single aggregate metric to measure model separability.
* Cons:
    * Doesn't indicate specific operating thresholds or class probability calibration quality.

### Visualizing AUC Area Representation

```python
import numpy as np
import matplotlib.pyplot as plt

fpr = np.linspace(0, 1, 100)
tpr = fpr ** 0.3 # Simulated convex ROC curve

plt.figure(figsize=(6, 3.5))
plt.plot(fpr, tpr, color='darkcyan', linewidth=2, label="AUC = 0.88")
plt.fill_between(fpr, tpr, color='darkcyan', alpha=0.2)
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')
plt.title("Area Under the Curve (AUC)", fontsize=11, fontweight='bold')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Confusion Matrix
* Definition: A tabular layout mapping actual target classes against model predictions.
* Use Cases:
    * Detailed breakdown and error analysis of classification outcomes.
* Pros:
    * Explicitly exposes exact types of errors (false positives vs. false negatives).
* Cons:
    * Difficult to interpret visually for multi-class problems with a massive number of labels without heatmaps.

### Visualizing Confusion Matrix Heatmap

```python
import numpy as np
import matplotlib.pyplot as plt

matrix = np.array([[45, 5], [3, 47]])

plt.figure(figsize=(5, 3.5))
plt.imshow(matrix, cmap='Blues', interpolation='nearest')
plt.title("Confusion Matrix Heatmap", fontsize=11, fontweight='bold')
plt.colorbar()
plt.xticks([0, 1], ['Pred 0', 'Pred 1'])
plt.yticks([0, 1], ['Actual 0', 'Actual 1'])

for i in range(2):
    for j in range(2):
        plt.text(j, i, str(matrix[i, j]), ha='center', va='center', color='black', fontweight='bold')

plt.tight_layout()
plt.show()
```

# Regression Metrics

## 1. Mean Squared Error (MSE)
* Definition: The average squared difference between estimated values and actual targets.
* Use Cases:
    * Evaluating regression models where large errors must be penalized heavily.
* Pros:
    * Differentiable everywhere and strongly penalizes large mistakes.
* Cons:
    * Highly sensitive to outliers; units are squared.

### Visualizing MSE Error Growth

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
mse = errors ** 2

plt.figure(figsize=(6, 3.5))
plt.plot(errors, mse, color='blue', linewidth=2)
plt.title("Mean Squared Error (MSE)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("Squared Error Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 2. Root Mean Squared Error (RMSE)
* Definition: The square root of the Mean Squared Error.
* Use Cases:
    * Regression evaluation where error needs to match the target variable's scale.
* Pros:
    * Interpretable in the exact same units as the target variable.
* Cons:
    * Retains the exact same outlier sensitivity as MSE.

### Visualizing RMSE Metric Scale

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
rmse = np.sqrt(errors ** 2)

plt.figure(figsize=(6, 3.5))
plt.plot(errors, rmse, color='darkblue', linewidth=2)
plt.title("Root Mean Squared Error (RMSE)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("RMSE Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 3. Mean Absolute Error (MAE)
* Definition: The average absolute magnitude of errors without squaring them.
* Use Cases:
    * Evaluating regression models on noisy datasets containing extreme outliers.
* Pros:
    * Robust and resistant to extreme outlier distortions.
* Cons:
    * Non-differentiable at zero.

### Visualizing MAE Linear Error Penalty

```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.linspace(-3, 3, 100)
mae = np.abs(errors)

plt.figure(figsize=(6, 3.5))
plt.plot(errors, mae, color='green', linewidth=2)
plt.title("Mean Absolute Error (MAE)", fontsize=11, fontweight='bold')
plt.xlabel(r"Prediction Error ($y - \hat{y}$)")
plt.ylabel("Absolute Error Value")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

## 4. R² Score (Coefficient of Determination)
* Definition: The proportion of variance in the dependent variable predictable from the independent variables ($1 - \frac{\text{SS}_{res}}{\text{SS}_{tot}}$).
* Use Cases:
    * Assessing how well a regression model fits relative to a simple baseline mean.
* Pros:
    * Normalized scale (typically up to 1.0) making model fit comparisons intuitive.
* Cons:
    * Can be misleading if evaluated on datasets without a meaningful mean baseline.

### Visualizing R² Score Variance Fit

```python
import numpy as np
import matplotlib.pyplot as plt

models = ['Baseline Mean', 'Poor Model', 'Good Model']
r2_scores = [0.0, 0.45, 0.91]

plt.figure(figsize=(6, 3.5))
plt.bar(models, r2_scores, color=['gray', 'orange', 'royalblue'], alpha=0.8, edgecolor='black')
plt.ylim(0, 1.0)
plt.title("R² Score Comparison", fontsize=11, fontweight='bold')
plt.ylabel("R² Score Value")
plt.grid(True, axis='y', linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
